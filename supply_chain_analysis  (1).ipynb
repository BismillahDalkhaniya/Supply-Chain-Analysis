### Outlier detection

```python
Q1 = df["Shipping costs"].quantile(0.25)
Q3 = df["Shipping costs"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Shipping costs"] < lower) |
    (df["Shipping costs"] > upper)
]

outliers
```

### 2. Product Performance

Question:

Which product types generate the most revenue?


```python
product_revenue = (
    df.groupby("Product type")["Revenue generated"]
    .sum()
    .sort_values(ascending=False)
)

product_revenue
```

```python
product_revenue.plot(kind="bar")

plt.title("Revenue by Product Type")
plt.xlabel("Product Type")
plt.ylabel("Revenue")
plt.show()
```

### Supplier Analysis

We want questions such as:

Which suppliers have longer lead times?

Which suppliers have higher defect rates?

Which suppliers have higher manufacturing costs?

```python
supplier_analysis = df.groupby("Supplier name").agg({
    "Lead time": "mean",
    "Defect rates": "mean",
    "Manufacturing costs": "mean"
})

supplier_analysis
```

### Correlation Analysis
### A better approach: make one correlation matrix

-The correlation coefficient r ranges from -1 to +1.

- Closer to +1 → stronger positive relationship
- Closer to -1 → stronger negative relationship
- Closer to 0 → weak/no linear relationship

```python
numeric_df = df.select_dtypes(include="number")

correlation_matrix = numeric_df.corr()

correlation_matrix
```

### Create calculated business metrics

#### - Revenue per unit

```python
df["revenue_per_unit"] = (
    df["Revenue generated"] /
    df["Number of products sold"]
)
df
```

#### - Shipping cost as % of revenue

#### - Manufacturing cost as % of revenue

```python
df["Manufacturing cost %"] = (
    df["Manufacturing costs"] /
    df["Revenue generated"]
) * 100
```

```python
df[["Manufacturing costs", "Revenue generated", "Manufacturing cost %"]].head()
```

```python
df["Shipping costs %"] = (
    df["Shipping costs"] /
    df["Revenue generated"]
) * 100
```
